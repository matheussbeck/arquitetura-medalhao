```python
# ============================================================================# VIOLIN PLOT# ============================================================================# Dados: TCH de 3 variedadesnp.random.seed(42)df = pd.DataFrame({    'variedade': ['RB867515']*50 + ['RB966928']*50 + ['CTC4']*50,    'tch': np.concatenate([        np.random.normal(85, 5, 50),   # RB867515: média 85, desvio 5        np.random.normal(90, 4, 50),   # RB966928: média 90, desvio 4        np.random.normal(88, 6, 50)    # CTC4: média 88, desvio 6    ])})fig = go.Figure()cores = [PRIMARY_COLOR, SUCCESS_COLOR, WARNING_COLOR]for i, variedade in enumerate(['RB867515', 'RB966928', 'CTC4']):    df_var = df[df['variedade'] == variedade]    fig.add_trace(go.Violin(        y=df_var['tch'],        name=variedade,        box_visible=True,          # Mostra box plot interno        meanline_visible=True,     # Mostra linha de média        fillcolor=cores[i],        opacity=0.6,        line_color=cores[i],        hovertemplate='<b>%{fullData.name}</b><br>TCH: %{y:.1f} t/ha<extra></extra>'    ))fig.update_layout(    title="Distribuição de TCH por Variedade (Violin Plot)",    yaxis_title="TCH (t/ha)",    plot_bgcolor='white',    paper_bgcolor='white',    showlegend=True,    font=dict(family="Segoe UI"))fig.show()
```

```python
# ============================================================================# ANÁLISE DE PARETO (REGRA 80/20)# ============================================================================from plotly.subplots import make_subplots# Dados: Área por variedadedf = pd.DataFrame({    'variedade': ['RB867515', 'RB966928', 'CTC4', 'SP81-3250',                  'RB92579', 'CTC20', 'RB975935', 'IACSP95'],    'area_ha': [3500, 2800, 2200, 1500, 1000, 800, 600, 400]})# Ordena do maior para o menordf = df.sort_values('area_ha', ascending=False).reset_index(drop=True)# Calcula acumuladodf['area_acum'] = df['area_ha'].cumsum()df['pct_acum'] = (df['area_acum'] / df['area_ha'].sum() * 100).round(1)# Identifica quais atingem 80%df['pareto_80'] = df['pct_acum'] <= 80# Cria gráfico com eixo duplofig = make_subplots(specs=[[{"secondary_y": True}]])# Barras: Área individualfig.add_trace(    go.Bar(        x=df['variedade'],        y=df['area_ha'],        name='Área (ha)',        marker_color=[SUCCESS_COLOR if p else '#e0e0e0' for p in df['pareto_80']],        text=df['area_ha'],        texttemplate='%{text:,.0f}',        textposition='outside',        hovertemplate='<b>%{x}</b><br>Área: %{y:,.0f} ha<extra></extra>'    ),    secondary_y=False)# Linha: Percentual acumuladofig.add_trace(    go.Scatter(        x=df['variedade'],        y=df['pct_acum'],        name='% Acumulado',        mode='lines+markers',        line=dict(color=DANGER_COLOR, width=3),        marker=dict(size=10, symbol='diamond'),        text=df['pct_acum'],        texttemplate='%{text}%',        textposition='top center',        hovertemplate='<b>%{x}</b><br>Acumulado: %{y:.1f}%<extra></extra>'    ),    secondary_y=True)# Linha de 80% (Regra de Pareto)fig.add_hline(    y=80,    line_dash="dash",    line_color="black",    line_width=2,    annotation_text="Regra 80/20",    annotation_position="right",    secondary_y=True)# Configurar eixosfig.update_xaxes(title_text="Variedade")fig.update_yaxes(title_text="Área (ha)", secondary_y=False)fig.update_yaxes(title_text="% Acumulado", range=[0, 105], secondary_y=True)fig.update_layout(    title="Análise de Pareto - Variedades por Área",    plot_bgcolor='white',    paper_bgcolor='white',    height=500,    hovermode='x',    showlegend=True,    legend=dict(orientation="h", y=-0.15, x=0.5, xanchor='center'),    font=dict(family="Segoe UI"))fig.show()
```

---## 5.3 Template de Tabelas### Tabela com Formatação Condicional

```python
# ============================================================================# TEMPLATE: TABELA COM FORMATAÇÃO CONDICIONAL# ============================================================================def criar_tabela_condicional(df, colunas_config):    """    Cria tabela Dash com formatação condicional.    Args:        df: DataFrame com os dados        colunas_config: Lista de dicts com configuração de cada coluna    Exemplo de colunas_config:        [            {                'id': 'talhao',                'name': 'Talhão',                'type': 'text'            },            {                'id': 'tch',                'name': 'TCH (t/ha)',                'type': 'numeric',                'format': {'specifier': '.1f'},                'condicoes': [                    {'operador': '<', 'valor': 80, 'cor': DANGER_COLOR},                    {'operador': '>=', 'valor': 90, 'cor': SUCCESS_COLOR}                ]            }        ]    Returns:        dash_table.DataTable    """    # Prepara colunas    columns = []    for col in colunas_config:        col_def = {            'name': col['name'],            'id': col['id'],            'type': col.get('type', 'text')        }        if 'format' in col:            col_def['format'] = col['format']        columns.append(col_def)    # Prepara formatação condicional    style_data_conditional = []    for col in colunas_config:        if 'condicoes' in col:            for cond in col['condicoes']:                style_data_conditional.append({                    'if': {                        'filter_query': f"{{{col['id']}}} {cond['operador']} {cond['valor']}",                        'column_id': col['id']                    },                    'backgroundColor': f"{cond['cor']}20",  # 20 = 12% transparência                    'color': cond['cor'],                    'fontWeight': 'bold'                })    # Adiciona zebrado    style_data_conditional.append({        'if': {'row_index': 'odd'},        'backgroundColor': '#f8f9fa'    })    return dash_table.DataTable(        data=df.to_dict('records'),        columns=columns,        style_table={'overflowX': 'auto'},        style_header={            'backgroundColor': PRIMARY_COLOR,            'color': 'white',            'fontWeight': 'bold',            'textAlign': 'center',            'padding': '12px'        },        style_cell={            'textAlign': 'center',            'padding': '12px',            'fontFamily': 'Segoe UI',            'fontSize': '13px'        },        style_data_conditional=style_data_conditional,        page_size=10,        sort_action='native',       # Permite ordenar        filter_action='native'      # Permite filtrar    )# ────────────────────────────────────────────────────────────────────────────# EXEMPLO DE USO# ────────────────────────────────────────────────────────────────────────────df = pd.DataFrame({    'talhao': ['T001', 'T002', 'T003', 'T004', 'T005'],    'tch': [92, 75, 88, 65, 95],    'atr': [148, 142, 146, 140, 152],    'area_ha': [45, 28, 38, 22, 50]})tabela = criar_tabela_condicional(    df,    colunas_config=[        {            'id': 'talhao',            'name': 'Talhão',            'type': 'text'        },        {            'id': 'tch',            'name': 'TCH (t/ha)',            'type': 'numeric',            'format': {'specifier': '.1f'},            'condicoes': [                {'operador': '<', 'valor': 80, 'cor': DANGER_COLOR},                {'operador': '>=', 'valor': 90, 'cor': SUCCESS_COLOR}            ]        },        {            'id': 'atr',            'name': 'ATR (kg/t)',            'type': 'numeric',            'format': {'specifier': '.1f'},            'condicoes': [                {'operador': '<', 'valor': 145, 'cor': WARNING_COLOR},                {'operador': '>=', 'valor': 150, 'cor': SUCCESS_COLOR}            ]        },        {            'id': 'area_ha',            'name': 'Área (ha)',            'type': 'numeric',            'format': {'specifier': ',.0f'}        }    ])# Usar no layout:html.Div([tabela])
```

---## 6.3 Títulos e Legendas### Anatomia de um Título Perfeito

```python
# ============================================================================# BOAS PRÁTICAS: TÍTULOS# ============================================================================# ❌ RUIM: Vago, sem contextotitulo_ruim = "Dados"# ⚠️ MÉDIO: Descritivo, mas sem insighttitulo_medio = "Área Plantada por Mês"# ✅ BOM: Descritivo + unidade + períodotitulo_bom = "Área Plantada por Mês (ha) - Safra 2023/2024"# 🌟 EXCELENTE: Descritivo + insight + call to actiontitulo_excelente = "Área Plantada Superou Meta em 12% - Safra 2023/2024"# ────────────────────────────────────────────────────────────────────────────# TEMPLATE DE TÍTULO COMPLETO# ────────────────────────────────────────────────────────────────────────────def criar_titulo_completo(principal, subtitulo=None, contexto=None):    """    Cria título rico com subtítulo e contexto.    Args:        principal: Título principal        subtitulo: Subtítulo opcional (ex: período, unidade)        contexto: Contexto adicional (ex: fonte, atualização)    Returns:        str: Título formatado em HTML    """    html_parts = [f"<b>{principal}</b>"]    if subtitulo:        html_parts.append(f"<br><sub>{subtitulo}</sub>")    if contexto:        html_parts.append(f"<br><sub style='color:#888'>{contexto}</sub>")    return ''.join(html_parts)# Exemplo de uso:fig.update_layout(    title={        'text': criar_titulo_completo(            principal="Evolução de TCH por Variedade",            subtitulo="Toneladas de Cana por Hectare - Safra 2023/2024",            contexto=f"Atualizado em {datetime.now().strftime('%d/%m/%Y')}"        ),        'font': {'size': 18, 'color': TEXT_DARK},        'x': 0.5,          # Centralizado        'xanchor': 'center'    })
```
