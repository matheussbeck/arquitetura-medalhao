---# <a id='ex1'></a>🔰 EXEMPLO 1: Anatomia de um Gráfico Plotly## 📚 O que você vai aprender:- Estrutura completa de um gráfico: **Figure, Trace, Layout, Config**- Como funciona cada componente- Passo a passo para criar qualquer gráfico## 🎯 Quando usar:Este é o FUNDAMENTO de tudo! Entenda bem este exemplo antes dos outros.## 🏗️ Arquitetura Visual:```┌─────────────────────────────────────┐│      Figure (Figura Completa)       ││  ┌───────────────────────────────┐  ││  │   Layout (Aparência)          │  │ ← Títulos, eixos, cores│  └───────────────────────────────┘  ││  ┌───────────────────────────────┐  ││  │   Trace 1 (Dados + Estilo)    │  │ ← Linha, barra, scatter│  ├───────────────────────────────┤  ││  │   Trace 2 (Dados + Estilo)    │  │ ← Múltiplas séries│  └───────────────────────────────┘  │└─────────────────────────────────────┘```## 💻 Código Explicado:

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS: Produtividade Mensal (TCH = Toneladas de Cana por Hectare)# ──────────────────────────────────────────────────────────────────────────meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun']tch = [85.2, 88.5, 92.3, 90.1, 87.8, 89.5]# ──────────────────────────────────────────────────────────────────────────# PASSO 1: Criar FIGURE (recipiente vazio)# ──────────────────────────────────────────────────────────────────────────fig = go.Figure()# ──────────────────────────────────────────────────────────────────────────# PASSO 2: Adicionar TRACE (dados + estilo visual)# ──────────────────────────────────────────────────────────────────────────fig.add_trace(    go.Scatter(                          # Tipo: gráfico de linha/pontos        x=meses,                         # Eixo X: categorias        y=tch,                           # Eixo Y: valores        mode='lines+markers',            # Linha COM marcadores        name='TCH Médio',                # Nome (legenda)                # Estilo da LINHA        line=dict(            color='#2E7D32',             # Verde escuro            width=3                      # Espessura        ),                # Estilo dos MARCADORES (pontos)        marker=dict(            size=10,                     # Tamanho            color='#66BB6A',             # Verde claro            line=dict(                   # Borda do ponto                color='#1B5E20',         # Verde mais escuro                width=2            )        ),                # Tooltip (hover)        hovertemplate='<b>%{x}</b><br>TCH: %{y:.1f} t/ha<extra></extra>'        # %{x} = valor X | %{y:.1f} = valor Y com 1 decimal    ))# ──────────────────────────────────────────────────────────────────────────# PASSO 3: Configurar LAYOUT (aparência geral)# ──────────────────────────────────────────────────────────────────────────fig.update_layout(    # Título    title=dict(        text='Produtividade Mensal (TCH)',        font=dict(size=20, color='#1B5E20'),        x=0.5,                           # Centraliza (0=esquerda, 1=direita)        xanchor='center'    ),        # Eixo X    xaxis=dict(        title='Mês',        showgrid=True,        gridcolor='rgba(0,0,0,0.1)'      # Cinza claro transparente    ),        # Eixo Y    yaxis=dict(        title='TCH (t/ha)',        showgrid=True,        gridcolor='rgba(0,0,0,0.1)',        range=[80, 95]                   # Escala fixa    ),        # Cores de fundo    plot_bgcolor='white',                # Área do gráfico    paper_bgcolor='#F5F5F5',             # Área externa        # Tamanho    width=800,    height=500)# ──────────────────────────────────────────────────────────────────────────# PASSO 4: Exibir gráfico# ──────────────────────────────────────────────────────────────────────────fig.show()print("\n✅ Gráfico criado!")print("\n🔍 Interaja:")print("   • Passe o mouse sobre os pontos")print("   • Arraste para dar zoom")print("   • Clique duplo para resetar")
````

# <a id='ex2'></a>🔰 EXEMPLO 2: Sistema de Cores Semânticas## 📚 O que você vai aprender:- Usar cores com **SIGNIFICADO** (não apenas decorativas)- Paleta completa para agronegócio- Quando usar verde/vermelho/amarelo/azul## 🎯 Quando usar:Cores semânticas comunicam informação instantaneamente:- 🟢 **Verde**: Positivo, meta atingida, crescimento- 🔴 **Vermelho**: Negativo, alerta, fora da meta- 🟡 **Amarelo**: Atenção, intermediário- 🔵 **Azul**: Informação neutra## 💡 Exemplo Prático:TCH por variedade colorido por performance (verde ≥85, amarelo 75-85, vermelho <75)

````python
# ──────────────────────────────────────────────────────────────────────────# PALETA DE CORES SEMÂNTICAS (Copy & Paste em seus projetos!)# ──────────────────────────────────────────────────────────────────────────COLORS = {    'success': '#4CAF50',    # ✅ Meta atingida    'warning': '#FF9800',    # ⚠️  Atenção    'danger': '#F44336',     # 🔴 Crítico    'info': '#2196F3',       # ℹ️  Neutro    'primary': '#2E7D32',    # Verde escuro (agro)}# ──────────────────────────────────────────────────────────────────────────# DADOS: TCH por Variedade# ──────────────────────────────────────────────────────────────────────────variedades = ['RB867515', 'RB966928', 'CTC4', 'SP81-3250', 'RB92579']tch = [88.5, 72.3, 92.1, 78.4, 81.2]# Meta: TCH >= 85 (verde), entre 75-85 (amarelo), < 75 (vermelho)META_ALTA, META_BAIXA = 85, 75# ──────────────────────────────────────────────────────────────────────────# FUNÇÃO: Determina cor baseada na performance# ──────────────────────────────────────────────────────────────────────────def get_color_by_performance(valor):    """Retorna cor semântica baseada no valor."""    if valor >= META_ALTA:        return COLORS['success']    # Verde: atingiu meta    elif valor >= META_BAIXA:        return COLORS['warning']    # Amarelo: atenção    else:        return COLORS['danger']     # Vermelho: crítico# Aplica corescores = [get_color_by_performance(t) for t in tch]# ──────────────────────────────────────────────────────────────────────────# GRÁFICO# ──────────────────────────────────────────────────────────────────────────fig = go.Figure()fig.add_trace(go.Bar(    x=variedades,    y=tch,    marker=dict(        color=cores,                    # 📌 Cada barra com cor semântica        line=dict(color='white', width=2)    ),    text=[f"{t:.1f}" for t in tch],    textposition='outside',    hovertemplate='<b>%{x}</b><br>TCH: %{y:.1f} t/ha<extra></extra>'))# Linhas de referência (metas)fig.add_hline(y=85, line_dash='dash', line_color=COLORS['success'],              annotation_text='Meta: 85 t/ha', annotation_position='right')fig.add_hline(y=75, line_dash='dash', line_color=COLORS['danger'],              annotation_text='Mínimo: 75 t/ha', annotation_position='right')fig.update_layout(    title='TCH por Variedade - Performance vs Meta',    xaxis_title='Variedade',    yaxis=dict(title='TCH (t/ha)', range=[0, 100]),    plot_bgcolor='white',    showlegend=False,    # Cores são autoexplicativas    width=800, height=500)fig.show()# ──────────────────────────────────────────────────────────────────────────# ANÁLISE AUTOMÁTICA# ──────────────────────────────────────────────────────────────────────────print("\n📊 ANÁLISE DE PERFORMANCE:")for var, t in zip(variedades, tch):    if t >= META_ALTA:        print(f"   ✅ {var}: {t:.1f} t/ha → EXCELENTE (meta atingida)")    elif t >= META_BAIXA:        print(f"   ⚠️  {var}: {t:.1f} t/ha → ATENÇÃO (próximo da meta)")    else:        print(f"   🔴 {var}: {t:.1f} t/ha → CRÍTICO (abaixo do mínimo)")
````

---# <a id='ex5'></a>🔰 EXEMPLO 5: Recursos de Interatividade## 📚 O que você vai aprender:- Customizar tooltips (hover)- Configurar botões da modebar- Adicionar filtros interativos## 🎯 Recursos:Todos os gráficos Plotly são interativos por padrão, mas você pode personalizar!

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS# ──────────────────────────────────────────────────────────────────────────variedades = ['RB867515', 'RB966928', 'CTC4', 'SP81-3250']tch = [88.5, 92.1, 85.3, 78.4]atr = [148.2, 151.5, 145.8, 142.1]area = [450, 380, 520, 290]# ──────────────────────────────────────────────────────────────────────────# GRÁFICO COM TOOLTIP PERSONALIZADO# ──────────────────────────────────────────────────────────────────────────fig = go.Figure()# Adiciona scatter com tooltip RICO em informaçõesfig.add_trace(go.Scatter(    x=tch,    y=atr,    mode='markers+text',    marker=dict(        size=area,          # Tamanho proporcional à área        color=tch,          # Cor baseada em TCH        colorscale='Greens',        showscale=True,        colorbar=dict(title='TCH')    ),    text=variedades,        # Rótulo    textposition='top center',        # 📌 TOOLTIP PERSONALIZADO (HTML!)    hovertemplate=        '<b>%{text}</b><br>' +                    # Variedade em negrito        '<br>' +        'TCH: %{x:.1f} t/ha<br>' +                # TCH        'ATR: %{y:.1f} kg/t<br>' +                # ATR        'Área: %{marker.size} ha<br>' +           # Área        '<br>' +        '<i>Clique para detalhes</i>' +        '<extra></extra>',                        # Remove box extra        customdata=area,        # Dados customizados    name='Variedades'))# Layoutfig.update_layout(    title='Gráfico Interativo - Passe o mouse!',    xaxis=dict(title='TCH (t/ha)', range=[75, 95]),    yaxis=dict(title='ATR (kg/t)', range=[140, 155]),    plot_bgcolor='white',    width=800, height=600,        # 📌 CONFIGURAÇÃO DA MODEBAR (botões)    modebar=dict(        orientation='v',     # Vertical        bgcolor='rgba(255,255,255,0.8)',        # Botões disponíveis: zoom, pan, select, reset, etc.    ),        # 📌 MODO DE HOVER    hovermode='closest',     # Opções: 'closest', 'x', 'y', 'x unified')fig.show()print("\n🖱️  RECURSOS INTERATIVOS:")print("   ✓ Passe o mouse: veja tooltip personalizado")print("   ✓ Arraste: zoom em área")print("   ✓ Clique duplo: reset zoom")print("   ✓ Clique simples: seleciona ponto")print("   ✓ Botões: salvar PNG, zoom, pan, etc.")
````

---# <a id='ex6'></a>📊 EXEMPLO 6: Gráfico de Linha (Séries Temporais)## 📚 O que você vai aprender:- Plotar múltiplas séries temporais- Formatar eixo de datas- Adicionar área sombreada (range)## 🎯 Quando usar:- Evolução temporal (dias, meses, anos)- Comparar tendências- Séries contínuas

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS: TCH Atual vs Meta# ──────────────────────────────────────────────────────────────────────────tch_atual = 87.5tch_meta = 90.0tch_minimo = 75.0tch_maximo = 105.0# ──────────────────────────────────────────────────────────────────────────# GAUGE COM FAIXAS DE CORES# ──────────────────────────────────────────────────────────────────────────fig = go.Figure(go.Indicator(    mode='gauge+number+delta',       # 📌 Modo: velocímetro + número + diferença    value=tch_atual,        # 📌 DELTA (diferença em relação à meta)    delta={        'reference': tch_meta,        'increasing': {'color': '#4CAF50'},  # Verde se maior        'decreasing': {'color': '#F44336'}   # Vermelho se menor    },        # 📌 TÍTULO    title={'text': 'TCH Médio Safra 2024<br><span style="font-size:14px">Toneladas por Hectare</span>'},        # 📌 CONFIGURAÇÃO DO GAUGE    gauge={        'axis': {            'range': [tch_minimo, tch_maximo],            'tickwidth': 1,            'tickcolor': '#333'        },                # 📌 FAIXAS DE CORES (semáforo)        'bar': {'color': '#2E7D32'},  # Cor da agulha        'steps': [            {'range': [tch_minimo, 80], 'color': '#FFCDD2'},  # Vermelho claro (crítico)            {'range': [80, 85], 'color': '#FFE0B2'},          # Laranja claro (atenção)            {'range': [85, tch_maximo], 'color': '#C8E6C9'}   # Verde claro (bom)        ],                # 📌 LINHA DE REFERÊNCIA (meta)        'threshold': {            'line': {'color': 'red', 'width': 4},            'thickness': 0.75,            'value': tch_meta        }    }))fig.update_layout(    width=600,    height=400,    font={'size': 16})fig.show()# ──────────────────────────────────────────────────────────────────────────# ANÁLISE# ──────────────────────────────────────────────────────────────────────────diferenca = tch_atual - tch_metapercentual = (diferenca / tch_meta) * 100print(f"\n📊 ANÁLISE DO KPI:")print(f"   TCH Atual: {tch_atual:.1f} t/ha")print(f"   Meta: {tch_meta:.1f} t/ha")print(f"   Diferença: {diferenca:+.1f} t/ha ({percentual:+.1f}%)")if diferenca >= 0:    print(f"   ✅ META ATINGIDA!")else:    print(f"   ⚠️  ABAIXO DA META")
````

---# <a id='ex16'></a>⚡ EXEMPLO 16: Waterfall Chart (Cascata)## 📚 O que você vai aprender:- Mostrar variações acumuladas- Visualizar contribuições positivas/negativas- Entender evolução passo a passo## 🎯 Quando usar:- Análise de variações (mês a mês, categoria a categoria)- Profit & Loss (P&L)- Explicar como chegou de A até B

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS: Variação do TCH ao Longo do Ano# ──────────────────────────────────────────────────────────────────────────categorias = [    'TCH Inicial (Jan)',     'Fev', 'Mar', 'Abr', 'Mai', 'Jun',    'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez',    'TCH Final (Dez)']valores = [    85,           # Inicial    +3, +4, -2, -3, +2,   # Variações mensais    -1, +2, +3, +2, -1, +1,    None          # Total será calculado automaticamente]# ──────────────────────────────────────────────────────────────────────────# WATERFALL CHART# ──────────────────────────────────────────────────────────────────────────fig = go.Figure(go.Waterfall(    x=categorias,        # 📌 MEDIDA: relative (variação) ou total (acumulado)    measure=['absolute'] + ['relative'] * 11 + ['total'],        y=valores,        # 📌 RÓTULOS NAS BARRAS    text=[f"{v:+.0f}" if v is not None and v != 85 else f"{v:.0f}" if v is not None else "" for v in valores],    textposition='outside',        # 📌 CORES    increasing={'marker': {'color': '#4CAF50'}},   # Verde (positivo)    decreasing={'marker': {'color': '#F44336'}},   # Vermelho (negativo)    totals={'marker': {'color': '#2196F3'}},       # Azul (total)        # 📌 CONECTORES (linhas que conectam as barras)    connector={'line': {'color': '#888', 'width': 2, 'dash': 'dot'}},        hovertemplate='%{x}<br>Variação: %{y:+.1f}<extra></extra>'))fig.update_layout(    title='Evolução do TCH - Análise de Variações Mensais (2024)',    yaxis=dict(title='TCH (t/ha)'),    xaxis=dict(tickangle=-45),    plot_bgcolor='white',    width=1000,    height=500,    showlegend=False)fig.show()print("\n📊 RESUMO:")tch_inicial = 85tch_final = tch_inicial + sum([v for v in valores[1:-1] if v is not None])print(f"   TCH Inicial (Jan): {tch_inicial:.1f} t/ha")print(f"   TCH Final (Dez): {tch_final:.1f} t/ha")print(f"   Variação Total: {tch_final - tch_inicial:+.1f} t/ha ({((tch_final/tch_inicial)-1)*100:+.1f}%)")print(f"\n   Meses positivos: {sum([1 for v in valores[1:-1] if v and v > 0])}")print(f"   Meses negativos: {sum([1 for v in valores[1:-1] if v and v < 0])}")
````

---# <a id='ex17'></a>⚡ EXEMPLO 17: Treemap Hierárquico## 📚 O que você vai aprender:- Visualizar hierarquias (fazenda > talhão > variedade)- Mostrar proporções em retângulos- Navegar em múltiplos níveis## 🎯 Quando usar:- Dados hierárquicos (região > fazenda > talhão)- Visualizar proporções de partes aninhadas- Substituir múltiplos gráficos de pizza

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS HIERÁRQUICOS: Fazenda → Talhão → Variedade# ──────────────────────────────────────────────────────────────────────────# Estrutura: (rótulo, pai, valor)labels = [    # Nível 0: Total    'Total',    # Nível 1: Fazendas    'Fazenda A', 'Fazenda B', 'Fazenda C',    # Nível 2: Variedades na Fazenda A    'FA - RB867515', 'FA - CTC4', 'FA - RB966928',    # Nível 2: Variedades na Fazenda B    'FB - RB867515', 'FB - CTC4',    # Nível 2: Variedades na Fazenda C    'FC - RB867515', 'FC - CTC4', 'FC - SP81-3250']parents = [    '',                                      # Total não tem pai    'Total', 'Total', 'Total',              # Fazendas → Total    'Fazenda A', 'Fazenda A', 'Fazenda A',  # Variedades → Fazenda A    'Fazenda B', 'Fazenda B',               # Variedades → Fazenda B    'Fazenda C', 'Fazenda C', 'Fazenda C'   # Variedades → Fazenda C]values = [    0,                          # Total (soma automática)    0, 0, 0,                    # Fazendas (soma automática)    450, 380, 170,              # FA    520, 280,                   # FB    390, 310, 200               # FC]# ──────────────────────────────────────────────────────────────────────────# TREEMAP# ──────────────────────────────────────────────────────────────────────────fig = go.Figure(go.Treemap(    labels=labels,    parents=parents,    values=values,        # 📌 RÓTULOS    textinfo='label+value+percent parent',  # Mostra: nome + valor + % do pai    texttemplate='<b>%{label}</b><br>%{value} ha<br>%{percentParent}',        # 📌 CORES    marker=dict(        colorscale='Greens',        cmid=450,                           # Valor médio para escala de cores        line=dict(color='white', width=2)    ),        # 📌 MODO DE EMPACOTAMENTO    tiling=dict(        packing='squarify'                  # Algoritmo: squarify, binary, dice, slice    ),        hovertemplate='<b>%{label}</b><br>Área: %{value} ha<br>%{percentParent} do total<extra></extra>'))fig.update_layout(    title='Distribuição de Área - Fazendas → Variedades (Hierárquico)',    width=900,    height=600)fig.show()print("\n🌳 ESTRUTURA HIERÁRQUICA:")print("   Total")print("   ├── Fazenda A (1,000 ha)")print("   │   ├── RB867515: 450 ha (45%)")print("   │   ├── CTC4: 380 ha (38%)")print("   │   └── RB966928: 170 ha (17%)")print("   ├── Fazenda B (800 ha)")print("   │   ├── RB867515: 520 ha (65%)")print("   │   └── CTC4: 280 ha (35%)")print("   └── Fazenda C (900 ha)")print("       ├── RB867515: 390 ha (43%)")print("       ├── CTC4: 310 ha (34%)")print("       └── SP81-3250: 200 ha (22%)")
````

---# <a id='ex18'></a>⚡ EXEMPLO 18: Violin Plot (Violino)## 📚 O que você vai aprender:- Combinar box plot + densidade (KDE)- Visualizar forma da distribuição- Comparar distribuições de múltiplos grupos## 🎯 Quando usar:- Entender FORMA da distribuição (bimodal, assimétrica, etc.)- Box plot simples não mostra densidade- Comparar múltiplas distribuições

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS: TCH por Variedade (distribuições diferentes)# ──────────────────────────────────────────────────────────────────────────np.random.seed(42)# Variedade 1: Distribuição normalrb867515 = np.random.normal(88, 4, 100)# Variedade 2: Distribuição bimodal (2 picos)rb966928 = np.concatenate([    np.random.normal(85, 2, 50),    np.random.normal(95, 2, 50)])# Variedade 3: Distribuição assimétrica (cauda longa)ctc4 = np.random.gamma(shape=30, scale=3, size=100)# ──────────────────────────────────────────────────────────────────────────# VIOLIN PLOT# ──────────────────────────────────────────────────────────────────────────fig = go.Figure()dados = {    'RB867515': rb867515,    'RB966928': rb966928,    'CTC4': ctc4}cores = ['#2E7D32', '#1976D2', '#F57C00']for i, (var, valores) in enumerate(dados.items()):    fig.add_trace(go.Violin(        y=valores,        name=var,                # 📌 CONFIGURAÇÕES DO VIOLINO        box_visible=True,            # Mostra box plot interno        meanline_visible=True,       # Mostra linha da média        fillcolor=cores[i],        opacity=0.6,                # 📌 PONTOS        points='all',                # Mostra todos os pontos (ou 'outliers', False)        pointpos=-0.5,               # Posição dos pontos (-1.5 a 1.5)        jitter=0.3,                  # Dispersão horizontal dos pontos                hovertemplate='<b>%{fullData.name}</b><br>TCH: %{y:.1f}<extra></extra>'    ))fig.update_layout(    title='Distribuições de TCH por Variedade - Violin Plot',    yaxis=dict(title='TCH (t/ha)', range=[70, 105]),    plot_bgcolor='white',    width=800,    height=600,    showlegend=True)fig.show()print("\n🎻 INTERPRETAÇÃO DAS FORMAS:\n")print("RB867515:")print("   • Forma SIMÉTRICA (sino)")print("   • Distribuição NORMAL")print(f"   • Média: {rb867515.mean():.1f} ± {rb867515.std():.1f}")print("\nRB966928:")print("   • Forma BIMODAL (2 picos!)")print("   • Pode indicar 2 subgrupos (ex: 2 talhões diferentes)")print(f"   • Média: {rb966928.mean():.1f}")print("\nCTC4:")print("   • Forma ASSIMÉTRICA")print("   • Cauda longa à direita")print(f"   • Média: {ctc4.mean():.1f}")
````

---# <a id='ex19'></a>⚡ EXEMPLO 19: Radar Chart (Gráfico de Radar)## 📚 O que você vai aprender:- Comparar múltiplas dimensões simultaneamente- Visualizar performance vs benchmark- Identificar pontos fortes e fracos## 🎯 Quando usar:- Análise multidimensional (5-8 dimensões)- Comparar perfis (real vs ideal, fazenda A vs B)- KPIs balanceados (não apenas 1 métrica)

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS: Performance da Fazenda em Múltiplas Dimensões# ──────────────────────────────────────────────────────────────────────────categorias = ['TCH', 'ATR', 'Eficiência', 'Qualidade', 'Custo', 'Sustentabilidade']# Valores de 0-100 para cada dimensãofazenda_atual = [85, 78, 92, 88, 65, 90]benchmark = [90, 85, 90, 90, 80, 85]  # Padrão ideal# ──────────────────────────────────────────────────────────────────────────# RADAR CHART# ──────────────────────────────────────────────────────────────────────────fig = go.Figure()# 📌 BENCHMARK (área sombreada)fig.add_trace(go.Scatterpolar(    r=benchmark + [benchmark[0]],        # Fecha o polígono (repete primeiro valor)    theta=categorias + [categorias[0]],    fill='toself',    fillcolor='rgba(255, 152, 0, 0.2)',  # Laranja transparente    line=dict(color='#FF9800', width=2, dash='dash'),    name='Benchmark (Ideal)'))# 📌 PERFORMANCE ATUALfig.add_trace(go.Scatterpolar(    r=fazenda_atual + [fazenda_atual[0]],    theta=categorias + [categorias[0]],    fill='toself',    fillcolor='rgba(46, 125, 50, 0.3)',  # Verde transparente    line=dict(color='#2E7D32', width=3),    name='Fazenda Atual'))fig.update_layout(    polar=dict(        radialaxis=dict(            visible=True,            range=[0, 100],              # Escala 0-100            tickvals=[20, 40, 60, 80, 100],            ticktext=['20', '40', '60', '80', '100']        ),        angularaxis=dict(            tickfont=dict(size=12)        )    ),    title='Performance Multidimensional - Atual vs Benchmark',    showlegend=True,    width=700,    height=700)fig.show()# ──────────────────────────────────────────────────────────────────────────# ANÁLISE GAP# ──────────────────────────────────────────────────────────────────────────print("\n📊 ANÁLISE DE GAPS (Atual vs Benchmark):\n")for cat, atual, ideal in zip(categorias, fazenda_atual, benchmark):    gap = atual - ideal    status = '✅' if gap >= 0 else '⚠️'    print(f"{status} {cat:20s}: {atual:3d} vs {ideal:3d} → Gap: {gap:+3d}")    print("\n💡 CONCLUSÕES:")print("   Pontos Fortes: Eficiência, Sustentabilidade")print("   Pontos a Melhorar: Custo, ATR")print("   Gap Total Médio: {:.1f} pontos".format(sum([a-b for a,b in zip(fazenda_atual, benchmark)])/len(categorias)))
````

---# <a id='ex20'></a>⚡ EXEMPLO 20: Funnel Chart (Funil)## 📚 O que você vai aprender:- Visualizar funil de conversão/processo- Identificar onde há maior perda- Calcular taxas de conversão## 🎯 Quando usar:- Processos sequenciais com perda em cada etapa- Funil de vendas, recrutamento, produção- Identificar gargalos

````python
# ──────────────────────────────────────────────────────────────────────────# DADOS: Funil de Produção (da Moagem até o Açúcar)# ──────────────────────────────────────────────────────────────────────────estagios = [    'Cana Colhida',    'Cana Entregue (Após Perdas)',    'Cana Moída',    'Caldo Extraído',    'Açúcar Produzido',    'Açúcar Empacotado']valores = [100000, 97500, 96000, 85000, 12500, 12000]  # toneladas# ──────────────────────────────────────────────────────────────────────────# CALCULA TAXAS DE CONVERSÃO# ──────────────────────────────────────────────────────────────────────────taxas = []for i in range(1, len(valores)):    taxa = (valores[i] / valores[i-1]) * 100    perda = 100 - taxa    taxas.append((taxa, perda))# ──────────────────────────────────────────────────────────────────────────# FUNNEL CHART# ──────────────────────────────────────────────────────────────────────────fig = go.Figure()fig.add_trace(go.Funnel(    y=estagios,    x=valores,        # 📌 RÓTULOS PERSONALIZADOS    textinfo='value+percent previous',  # Mostra valor + % do anterior    texttemplate='<b>%{y}</b><br>%{x:,.0f} t<br>(%{percentPrevious})',    textposition='inside',    textfont=dict(size=11, color='white'),        # 📌 CORES (gradiente verde)    marker=dict(        color=['#1B5E20', '#2E7D32', '#388E3C', '#4CAF50', '#66BB6A', '#81C784']    ),        # 📌 CONECTORES    connector=dict(        line=dict(color='#888', width=2)    ),        hovertemplate='<b>%{y}</b><br>Quantidade: %{x:,.0f} t<br>%{percentPrevious} da etapa anterior<extra></extra>'))fig.update_layout(    title='Funil de Produção - Da Colheita ao Açúcar Empacotado',    width=800,    height=600,    showlegend=False)fig.show()# ──────────────────────────────────────────────────────────────────────────# ANÁLISE DE PERDAS# ──────────────────────────────────────────────────────────────────────────print("\n📊 ANÁLISE DO FUNIL:\n")for i, (estagio, valor) in enumerate(zip(estagios, valores)):    pct_inicial = (valor / valores[0]) * 100    print(f"{i+1}. {estagio:35s}: {valor:>8,.0f} t ({pct_inicial:5.1f}% do total)")    if i < len(taxas):        taxa, perda = taxas[i]        print(f"   → Conversão para próxima etapa: {taxa:.1f}% (perda: {perda:.1f}%)\n")print(f"\n💡 EFICIÊNCIA GLOBAL: {(valores[-1]/valores[0])*100:.1f}%")print(f"   De {valores[0]:,.0f}t de cana → {valores[-1]:,.0f}t de açúcar empacotado")
````
